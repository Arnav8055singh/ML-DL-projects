Importing Kaggle DataSet

```python
import kagglehub


path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

print("Path to dataset files:", path)
```

Import The Dependecies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
#load the dataset to a Pandas Dataframe

credit_card_data = pd.read_csv('/content/creditcard.csv')
```

```python
#first 5 rows of the dataset

credit_card_data.head()
```

```python
credit_card_data.tail()
```

```python
#dataset information

credit_card_data.info()
```

```python
#checking the missing values in each column

credit_card_data.isnull().sum()
```

```python
#Distribution of legit Transactions and fraudulent Transactions
credit_card_data['Class'].value_counts()
```

The DataSet is Highly Imbalanced
0-> Normal Transactions
1-> Fraudelent Transactions


```python
#Separating the Data For Analaysis

legit = credit_card_data [credit_card_data.Class == 0]
fraud = credit_card_data[credit_card_data.Class == 1]


```

```python
print(legit.shape)
print(fraud.shape)
```

```python
#Statistical Measures for the data

legit.Amount.describe()
```

```python
fraud.Amount.describe()
```

```python
#compare the values for both Transactions

credit_card_data.groupby('Class').mean()
```

Under Sampling



Number of Fraudelent Transactions --> 103


```python
legit_sample = legit.sample(n=103)
```

Cocatenating two DataFrames


```python
new_dataset = pd.concat([legit_sample, fraud], axis=0)
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

Spliting the Data into Features And Targets

```python
X = new_dataset.drop(columns='Class', axis=1)
Y = new_dataset['Class']
```

```python
print(X)
```

```python
print(Y)
```

Split the Data into Training data and the testing data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, stratify=Y, random_state=2)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training -> Logistic Regression

```python
model = LogisticRegression()
```

```python
#Training the logistic Regression Model with Training Data
```

```python
model.fit(X_train, Y_train)
```

Evaluation of Model with Accuracy Model

```python
X_train_prediction = model.predict(X_train)
training_data_accuracy = accuracy_score(X_train_prediction, Y_train)
```

```python
print('Accuracy on training data : ', training_data_accuracy)
```

Getting the Accuracy of 97.5%

```python
#Accuracy on Test data

X_test_prediction = model.predict(X_test)
test_data_accuracy = accuracy_score(X_test_prediction, Y_test)
```

```python
print('Accuracy score on Test Data: ', test_data_accuracy)
```
